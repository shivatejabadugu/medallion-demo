```python
from pyspark.sql import SparkSession
spark = SparkSession.builder.appName("car_sales").getOrCreate()

df = spark.read.format('csv').option('inferSchema',True).option('header',True).load('C:/Users/shiva/Downloads/Car_sales.csv')
```

```python
df.limit(3).toPandas()
```

```python
from pyspark.sql.types import *
from pyspark.sql.functions import *
```

```python
#df_select = df.select('Manufacturer','Model','Unit Sales','Price')
df_select = df.select(col('Manufacturer'),col('Model'),col('Unit Sales'),col('Price'))

df_select.limit(5).toPandas()
```

```python
df_alias = df_select(col('Unit Sales').alias('Unit_Sales'))
df_alias.show()
```

```python
df.show()
```

Filter

```python
#df.filter((col('Manufacturer')=='Audi')).show()
df_filter = df.filter((col('Manufacturer')=='Audi') & (col('Retention Value') == 'GOOD'))

df_filter.toPandas()
```

```python
df.show()
```

```python
df_multi_filter = df.filter(
    (col('Horsepower').isNull()) & 
    (col('Retention Value').isin('GOOD', 'POOR'))
    )

df_multi_filter.toPandas()
```

```python
df.withColumnRenamed('Unit Sales','Unit_Sales').show(3)
```

```python
for c in ['Price', 'Year Resale Value']:
    df = df.withColumn(c, regexp_replace(col(c), '[^0-9.-]', '').cast('double'))

df.printSchema()
```

```python
df2 = df.withColumn('Price_Diff', 
                    col('Price') - col('Year Resale Value'))

df2.limit(3).toPandas()
```

```python
df_reg = df.withColumn('HP Level',regexp_replace(col('HP Level'),"Medium HP","MHP"))\
            .withColumn('HP Level',regexp_replace(col('HP Level'),"High HP","HHP"))

df_reg.limit(20).toPandas()
```
