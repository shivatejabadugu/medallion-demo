### DateFormat

```python
#df = df.withColumn('curr_date',current_date())

#df = df.withColumn('7DaysAfter',date_add('curr_date',7))
df = df.withColumn('7DaysBefore',date_add('curr_date',-7))
df.limit(3).toPandas()
```

```python
df_dates = df.withColumn('curr_date', current_date()) \
             .withColumn('7DaysAfter', date_add('curr_date', 7)) \
             .withColumn('diffdate', datediff('7DaysAfter', 'curr_date')) \
             .withColumn('curr_date', date_format('curr_date', 'dd-MM-yyyy')) \
             .withColumn('7DaysAfter', date_format('7DaysAfter', 'dd-MM-yyyy')) \
             .select('Manufacturer', 'Model', 'curr_date', '7DaysAfter', 'diffdate')

df_dates.limit(3).toPandas()
```

### Handling Nulls values
## Dropping Nulls

```python
df_null = df.dropna('any')

df_null.limit(5).toPandas()
```

```python
df.dropna('any')                              # drop rows with a null in ANY column (the default)
df.dropna(subset=['Horsepower'])              # drop rows only if Horsepower is null
df.dropna(thresh=15)                          # keep rows with at least 15 non-null values
```

```python
df.show()
print(df.count(), df_null.count())
```

```python
df_clean = df
for c in ['Price', 'Year Resale Value']:
    df_clean = df_clean.withColumn(c, regexp_replace(col(c), '[^0-9.-]', '').cast('double'))

df_clean.printSchema() 
```

```python
df_when1 = df_clean.withColumn('flag',
                 when (col('Year Resale Value')>50000,'Super good')
                .when((col('Year Resale Value') >= 40000) & (col('Year Resale Value') <= 50000), 'Good to buy')
                .when((col('Year Resale Value') >= 30000) & (col('Year Resale Value') < 40000), 'Average')
                .otherwise('Not Good to Buy')) \
                .withColumn('Depreciation',col('Price') - col('Year Resale Value'))\
                .select('Manufacturer', 'Model', 'Price', 'Year Resale Value','Depreciation', 'flag')

df_when1.show()
```

```python
df_clean.groupBy('Manufacturer') \
        .agg(avg('Price').alias('Total_Price')) \
        .orderBy(col('Total_Price').desc()) \
        .show()
```

```python
df_When = df.withColumn('Fav', when(col('Manufacturer') == 'Mercedes-B', 'like')\
            .when(col('Manufacturer')== 'Volvo','like')\
            .when(col('Manufacturer')== 'Porsche','Maybe')
            .otherwise('Dislike')) \
            .select('Manufacturer', 'Model', 'Horsepower', 'Fav')

df_When.limit(20).toPandas()
```
